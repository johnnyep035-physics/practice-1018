```python
import yfinance as yf
import pandas as pd
from datetime import datetime

# 設定股票代號（台積電在台灣證券交易所代號為 2330.TW）
ticker = "2330.TW"

# 設定時間範圍：2026 年初至今
start_date = "2026-01-01"
end_date = datetime.today().strftime('%Y-%m-%d')

# 抓取數據
df = yf.download(ticker, start=start_date, end=end_date)

# 整理欄位名稱以防 MultiIndex 格式或大小寫不一致
if not df.empty:
    if isinstance(df.columns, pd.MultiIndex):
        df.columns = df.columns.droplevel(1)

    # 顯示前幾筆數據
    display(df.head())
else:
    print("未找到 2026 年起的歷史數據，請確認時間是否已到 2026 年或該區間是否有交易資料。")
```

```python
import mplfinance as mpf

if not df.empty:
    # 繪製 K 線圖與成交量
    mpf.plot(
        df,
        type='candle',
        style='charles',
        title='TSMC (2330.TW) Daily Stock Price - 2026',
        ylabel='Price (TWD)',
        volume=True,
        ylabel_lower='Volume',
        figsize=(14, 8)
    )
else:
    print("無數據可繪製圖表。")
```

# 新增區段

```python
import matplotlib.pyplot as plt

if not df.empty:
    # 確保 Close 欄位是 1D Series，並計算移動平均線
    close_prices = df['Close']

    # 計算 5 日、10 日與 20 日移動平均線 (MA)
    df['MA5'] = close_prices.rolling(window=5).mean()
    df['MA10'] = close_prices.rolling(window=10).mean()
    df['MA20'] = close_prices.rolling(window=20).mean()

    # 顯示包含移動平均線的最新數據表格
    display(df[['Close', 'MA5', 'MA10', 'MA20']].tail(10))
else:
    print("目前沒有資料可計算移動平均線。")
```

```python
if not df.empty:
    # 繪製收盤價與移動平均線折線圖
    plt.figure(figsize=(14, 7))
    plt.plot(df.index, df['Close'], label='Close Price', color='blue', linewidth=2)
    plt.plot(df.index, df['MA5'], label='5-Day MA (MA5)', color='orange', linestyle='--')
    plt.plot(df.index, df['MA10'], label='10-Day MA (MA10)', color='green', linestyle='--')
    plt.plot(df.index, df['MA20'], label='20-Day MA (MA20)', color='red', linestyle='--')

    plt.title('TSMC (2330.TW) Daily Close Price and Moving Averages - 2026', fontsize=16)
    plt.xlabel('Date', fontsize=12)
    plt.ylabel('Price (TWD)', fontsize=12)
    plt.legend(loc='best')
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.gcf().autofmt_xdate() # 自動旋轉日期標籤避免重疊
    plt.show()
else:
    print("無數據可繪製圖表。")
```

```python
import plotly.graph_objects as go

if not df.empty:
    fig = go.Figure()

    # 新增收盤價折線
    fig.add_trace(go.Scatter(x=df.index, y=df['Close'], name='Close Price', line=dict(color='blue', width=2)))

    # 新增均線
    fig.add_trace(go.Scatter(x=df.index, y=df['MA5'], name='5-Day MA (MA5)', line=dict(color='orange', width=1.5, dash='dash')))
    fig.add_trace(go.Scatter(x=df.index, y=df['MA10'], name='10-Day MA (MA10)', line=dict(color='green', width=1.5, dash='dash')))
    fig.add_trace(go.Scatter(x=df.index, y=df['MA20'], name='20-Day MA (MA20)', line=dict(color='red', width=1.5, dash='dash')))

    # 設定圖表版面
    fig.update_layout(
        title='TSMC (2330.TW) Interactive Close Price & Moving Averages - 2026',
        xaxis_title='Date',
        yaxis_title='Price (TWD)',
        template='plotly_white',
        hovermode='x unified',
        legend=dict(yanchor="top", y=0.99, xanchor="left", x=0.01)
    )

    fig.show()
else:
    print("無數據可繪製互動式圖表。")
```

```python
if not df.empty:
    # 選擇核心的股價欄位進行描述性統計分析
    stats_df = df[['Open', 'High', 'Low', 'Close', 'Volume']].describe()

    print("--- 台積電 2026 年初至今股價數據描述性統計 ---")
    display(stats_df)
else:
    print("目前沒有資料可進行統計分析。")
```

```python
import yfinance as yf
import pandas as pd
from datetime import datetime

# 設定股票代號與新的時間範圍（從 2024 年初開始）
ticker = "2330.TW"
start_date_2024 = "2024-01-01"
end_date_today = datetime.today().strftime('%Y-%m-%d')

# 重新抓取數據
df_2024 = yf.download(ticker, start=start_date_2024, end=end_date_today)

# 整理欄位格式
if not df_2024.empty:
    if isinstance(df_2024.columns, pd.MultiIndex):
        df_2024.columns = df_2024.columns.droplevel(1)

    # 計算 5 日、10 日、20 日移動平均線
    df_2024['MA5'] = df_2024['Close'].rolling(window=5).mean()
    df_2024['MA10'] = df_2024['Close'].rolling(window=10).mean()
    df_2024['MA20'] = df_2024['Close'].rolling(window=20).mean()

    display(df_2024.head())
else:
    print("未能成功獲取 2024 年至今的數據。")
```

```python
import mplfinance as mpf

if not df_2024.empty:
    # 繪製 2024 至今的 K 線圖與成交量
    mpf.plot(
        df_2024,
        type='candle',
        style='charles',
        title='TSMC (2330.TW) Daily Stock Price - From 2024',
        ylabel='Price (TWD)',
        volume=True,
        ylabel_lower='Volume',
        figsize=(14, 8)
    )
else:
    print("無數據可繪製 K 線圖。")
```

```python
import plotly.graph_objects as go

if not df_2024.empty:
    fig_2024 = go.Figure()

    # 收盤價與均線折線
    fig_2024.add_trace(go.Scatter(x=df_2024.index, y=df_2024['Close'], name='Close Price', line=dict(color='blue', width=2)))
    fig_2024.add_trace(go.Scatter(x=df_2024.index, y=df_2024['MA5'], name='5-Day MA (MA5)', line=dict(color='orange', width=1.5, dash='dash')))
    fig_2024.add_trace(go.Scatter(x=df_2024.index, y=df_2024['MA10'], name='10-Day MA (MA10)', line=dict(color='green', width=1.5, dash='dash')))
    fig_2024.add_trace(go.Scatter(x=df_2024.index, y=df_2024['MA20'], name='20-Day MA (MA20)', line=dict(color='red', width=1.5, dash='dash')))

    fig_2024.update_layout(
        title='TSMC (2330.TW) Interactive Close Price & MAs - From 2024',
        xaxis_title='Date',
        yaxis_title='Price (TWD)',
        template='plotly_white',
        hovermode='x unified'
    )

    fig_2024.show()
else:
    print("無數據可繪製互動式圖表。")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 讀取 Iris 資料集
iris_df = pd.read_csv('/content/Iris.csv')

# 預覽資料結構
display(iris_df.head())

# 繪製 Pairwise Scatter Plot
# 排除可能存在的 Id 欄位，並以 'Species'（或您的分類欄位名稱）進行著色分類
plot_cols = [col for col in iris_df.columns if col.lower() != 'id']

# 偵測分類欄位名稱 (通常是 Species 或 class)
species_col = [col for col in iris_df.columns if 'species' in col.lower() or 'class' in col.lower()]

if species_col:
    sns.pairplot(iris_df[plot_cols], hue=species_col[0], palette='Set2', corner=True)
else:
    sns.pairplot(iris_df[plot_cols], corner=True)

plt.suptitle('Iris Dataset Pairwise Scatter Plot', y=1.02, fontsize=16)
plt.show()
```
